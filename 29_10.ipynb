```python
from IPython import get_ipython
from IPython.display import display
# %%

# %%
# Create a text file , enter data, Write operation
with open("students.txt","w") as file:
  file.write("Name, Department, Marks, City\n"),
  file.write("Ashu, CSE, 90, Gorakhpur\n"),
  file.write("Abhiyank, CSE, 85,Mathura\n"),
  file.write("Aditya, CSE, 88, Agra\n"),
  file.write("Ayansh, CSE, 84, Chaumuha")

# %%
with open("students.txt","r") as file:
  data =file.read()
  print(data)
# %% [markdown]
# ## Save the .txt data into .csv **bold text**
# %%
import pandas as pd

df=pd.read_csv("students.txt")
df.to_csv("students.csv",index=False)
print(df)
# %%
#Pickle
df_pickle=pd.DataFrame({
    "Name":["Ashu","Abhiyank","Aditya","Ayansh"],
    "Marks":[90,85,88,84]
})
df.to_pickle("students_pkl.pkl")
# %%
spl=pd.read_pickle("students_pkl.pkl")
print(spl)
# %%
import requests
from bs4 import BeautifulSoup
import pandas as pd
url="https://www.flipkart.com/"
page = requests.get(url)
soup=BeautifulSoup(page.text, "html.parser")
print(soup.title.text)
# %%
#parquet
df_parquet=df_pickle
df_parquet.to_parquet("students_pqt.parquet",index=False)
students_pqt=pd.read_parquet("students_pqt.parquet")
print(students_pqt)
# %%
#sqlite
import sqlite3
conn=sqlite3.connect("students.db")

cursor=conn.cursor()
print("DataBase created Successfully")
# %%
cursor.execute("CREATE TABLE IF NOT EXISTS students (name TEXT, Roll_no INTEGER, marks INTEGER)")
print("Table created Successfully!!")
# %%
student_data = [
    ('Ashu', 101, 90),
    ('Abhiyank', 102, 85),
    ('Aditya', 103, 88),
    ('Ayansh', 104, 84)
]

cursor.executemany("INSERT INTO students VALUES (?, ?, ?)", student_data)
# %%
conn = sqlite3.connect("students.db")
curso = conn.cursor()
cursor.execute(" SELECT * FROM students")
rows = cursor.fetchall()
for rows in rows:
  print(rows)
# %%

```
