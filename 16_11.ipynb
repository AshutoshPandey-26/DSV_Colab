```python
from io import StringIO
import pandas as pd

data = StringIO('''
Roll_No,Branch,Gender,Study_Hours,Attendance,Maths,Python,ML
1,ECE,Female,3.5,70,77,50,88
2,CSE,Female,5.0,84,82,78,94
3,ECE,Female,3.6,88,76,80,77
4,ECE,Male,3.4,71,67,69,82
5,CSE,Male,5.9,64,86,80,75
6,CSE,Male,3.6,69,81,72,64
7,ECE,Male,0.5,84,66,75,74
8,IT,Female,4.5,76,83,81,60
9,ECE,Male,3.4,83,72,74,61
10,ECE,Male,3.6,96,93,73,93
''')

df = pd.read_csv(data)

df["Total"] = df[["Maths", "Python", "ML"]].sum(axis=1)

df["Percentage"] = (df["Total"] / 3).round(1)

df["Grade"] = "D"

df.loc[df["Percentage"] >= 60, "Grade"] = "C"

df.loc[df["Percentage"] >= 70, "Grade"] = "B"

df.loc[df["Percentage"] >= 80, "Grade"] = "A"

df.head()

print(df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Group and count the number of students in each course (Branch)
branch_counts = df['Branch'].value_counts().reset_index()
branch_counts.columns = ['Branch', 'Student_Count']

# Create a bar graph
plt.figure(figsize=(8, 5))
sns.barplot(x='Branch', y='Student_Count', data=branch_counts, palette='viridis', hue='Branch', legend=False)

# Add labels and title
plt.title('Number of Students in Each Course', fontsize=14)
plt.xlabel('Course / Branch', fontsize=12)
plt.ylabel('Number of Students', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Show plot
plt.tight_layout()
plt.show()
```
