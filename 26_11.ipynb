```python
from IPython import get_ipython
from IPython.display import display
# %%
import pandas as pd
# create data frame
data={
    'date':['1/1/2017', '1/2/2017', '1/3/2017', '1/4/2017',
            '1/1/2017', '1/2/2017', '1/3/2017', '1/4/2017',
            '1/1/2017', '1/2/2017', '1/3/2017', '1/4/2017'],
    'city':['New York']*4 + ['Mumbai']*4 + ['Paris']*4,
    'temperature':[32,36,28,33,90,85,87,92,45,50,54,42],
    'Windspeed':[6,7,12,15,5,20,13,8,9,10,11,15],
    'event':['Rain', 'Sunny', 'Snow', 'Snow', 'Rain', 'Sunny', 'Snow', 'Snow', 'Rain', 'Sunny', 'Snow', 'Snow']
}
df=pd.DataFrame(data)
df
# %%
df1=df.groupby('city')
df1
# %%
for city, data in df1:
    print(city)
    print(data)
# %%
df2=df.groupby('event')
df2
# %%
for event, data in df2:
    print(event)
    print(data)
    
# %%
df1.get_group('New York')
# %%
df1.max()
# %%
df1.mean(numeric_only=True)
# %%
df1.aggregate('mean',numeric_only=True)
# %%
print("\n",df1.aggregate({
    'temperature':['min', 'max', 'mean'],
    'Windspeed':['min', 'max']
}))
# %%
result=df1.aggregate({
    'temperature':['min', 'max', 'mean'],
    'Windspeed':['min', 'max']
})
result
# %%
# row wise aggregation
print(df)
result=df[['temperature', 'Windspeed']].aggregate('sum', axis=0)
print(result)
# %%
# pivot table
table= df.pivot_table(index='city', values='temperature',aggfunc='mean')
print(table) #city wise temp avg
# %%
table= df.pivot_table(
    index='city',
    values=['temperature','Windspeed'],
    aggfunc=['sum','mean'],
    margins=True,
    )
print(table)
# %%
table= df.pivot_table(
    index='city',
    values=['temperature','Windspeed'],
    aggfunc='sum',
    margins=True,
    margins_name='Total'
    )
print(table)
# %%
result=pd.crosstab(df['city'], df['date'])
print(result) 
# %%

```
